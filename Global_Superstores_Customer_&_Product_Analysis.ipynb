# Question 2 Are there specific Products that sell in high quantities but generate low Profit?


```python
# Calculate profit per product
profit_per_product = df.groupby('Product.Name').agg({'Profit': 'sum', 'Sales': 'sum'}).reset_index()

# Filter for high sales and low profit
high_sales_low_profit = profit_per_product[(profit_per_product['Sales'] > profit_per_product['Sales'].median()) &
                                            (profit_per_product['Profit'] < profit_per_product['Profit'].median())]

# Sort and select top 20 products
top_20_high_sales_low_profit = high_sales_low_profit.sort_values(by='Sales', ascending=False).head(20)

# Display results
print(top_20_high_sales_low_profit)
```

```python
import matplotlib.pyplot as plt

# Top 20 by Profit
top_profit_products = df.groupby('Product.Name')['Profit'].sum().sort_values(ascending=False).head(20)

# Plotting the bar chart
plt.figure(figsize=(10, 6))
top_profit_products.plot(kind='bar', color='skyblue')
plt.title('Top 20 Products by Profit')
plt.xlabel('Product Name')
plt.ylabel('Total Profit')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Top 20 by Sales
top_sales_products = df.groupby('Product.Name')['Sales'].sum().sort_values(ascending=False).head(20)

# Plotting the bar chart
plt.figure(figsize=(10, 6))
top_sales_products.plot(kind='bar', color='green')
plt.title('Top 20 Products by Sales')
plt.xlabel('Product Name')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by Category
category_analysis = data.groupby(['Category']).agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()

# Group by Sub.Category
subcategory_analysis = data.groupby(['Sub.Category']).agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()

# Sort by Profit for better visualization
category_sorted = category_analysis.sort_values(by='Profit', ascending=False)
subcategory_sorted = subcategory_analysis.sort_values(by='Profit', ascending=False)

# Bar chart for Category Analysis
plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='Profit', data=category_sorted, palette='viridis')
plt.title('Profit by Category', fontsize=16)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Profit', fontsize=12)
plt.show()

# Bar chart for Sub.Category Analysis
plt.figure(figsize=(12, 8))
sns.barplot(y='Sub.Category', x='Profit', data=subcategory_sorted, palette='coolwarm')
plt.title('Profit by Sub-Category', fontsize=16)
plt.xlabel('Profit', fontsize=12)
plt.ylabel('Sub.Category', fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Aggregate data by Category and Sub.Category
combined_analysis = data.groupby(['Category', 'Sub.Category']).agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()

# Plotting
plt.figure(figsize=(10, 7))
sns.barplot(
    x='Profit',
    y='Sub.Category',
    hue='Category',
    data=combined_analysis,
    palette='Set2',
    orient='h'
)

# Add Titles and Labels
plt.title('Profit by Sub.Category within Each Category', fontsize=16)
plt.xlabel('Profit', fontsize=12)
plt.ylabel('Sub.Category', fontsize=12)
plt.legend(title='Category', fontsize=10)
plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Group data by Region and Market to get total sales
geo_data = data.groupby(['Region', 'Market'], as_index=False)['Sales'].sum()

# Normalize Sales for better color contrast
geo_data['Sales_normalized'] = (geo_data['Sales'] - geo_data['Sales'].min()) / (geo_data['Sales'].max() - geo_data['Sales'].min())

# Print the data to check values
print(geo_data)

# Create a choropleth map using Plotly
fig = px.choropleth(
    geo_data,
    locations='Region',  # Column with geographical names
    locationmode='USA-states',  # For U.S. state-level regions
    color='Sales_normalized',  # Normalized data for better color contrast
    hover_name='Market',  # Info to show on hover
    color_continuous_scale='Turbo',  # High-contrast color scale
)

# Update layout for map appearance
fig.update_geos(
    scope='usa',  # Limit the map to the USA
    projection_type='albers usa'  # Correctly set projection type
)

# Add margin adjustments
fig.update_layout(
    margin={"r": 0, "t": 50, "l": 0, "b": 0},
    title="Sales by Region and Market"
)

# Show the map
fig.show()
```
